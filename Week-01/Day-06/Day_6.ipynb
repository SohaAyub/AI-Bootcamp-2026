```python
marks = [85, 90, 78, 92, 88]

print("Student Marks:")
for mark in marks:
    print(mark)

print("\nTotal Marks:", sum(marks))
print("Average Marks:", sum(marks) / len(marks))
```

```python
shopping_list = ["Milk", "Bread", "Eggs"]

print("Original List:", shopping_list)

shopping_list.append("Butter")
print("After Adding Butter:", shopping_list)

shopping_list.remove("Bread")
print("After Removing Bread:", shopping_list)
```

```python
grades = [78, 85, 92, 67, 88, 95]

print("Student Grades:", grades)
print("Highest Grade:", max(grades))
print("Lowest Grade:", min(grades))
print("Total Students:", len(grades))
print("Average Grade:", sum(grades) / len(grades))
```
